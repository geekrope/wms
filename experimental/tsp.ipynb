```python
import numpy as np
import networkx as nx
import heapq
```

```python
def mst(vertices):
    n = vertices.shape[0]
    used = np.zeros(n).astype(bool)  
    used[0] = True  
    edges = []
    tree = []

    for u in range(1, n):
        edge_data = [np.linalg.norm(vertices[0] - vertices[u]), 0, u]
        heapq.heappush(edges, edge_data)
        
    while len(tree) < n - 1:
        edge = heapq.heappop(edges)
        v = edge[2]

        if used[v]: 
            continue
        
        used[v] = True        
        tree.append([edge[1], edge[2]])

        for u in range(n):
            if not used[u]:
                edge_data = [np.linalg.norm(vertices[v] - vertices[u]), v, u]
                heapq.heappush(edges, edge_data)                
    return tree

def preorder(current, parent, adjacency, result):
    result.append(current)
    for child in adjacency[current]:
        if child != parent:
            preorder(child, current, adjacency, result)
            # if we add the parent again, 
            # we can get a tour of the tree, 
            # which is bypassed using triangle 
            # inequality and going straight 
            # to the next vertex in traversal.
    return result
```

```python
n = 10

vertices = np.random.rand(n, 2)
graph = nx.complete_graph(vertices.shape[0])

nx.draw(graph, pos=vertices, with_labels=True)
```

```python
min_tree = mst(vertices)
min_tree_graph = nx.Graph(min_tree)

nx.draw(min_tree_graph, pos=vertices, with_labels=True)
```

```python
adjacency = [[] for i in range(n)] 
traversal = []
for edge in min_tree:
    adjacency[edge[0]].append(edge[1])
    adjacency[edge[1]].append(edge[0])

preorder(0, -1, adjacency, traversal)
print(traversal)

tsp_edges = list(zip(traversal[:-1], traversal[1:]))
tsp_edges.append((traversal[-1], traversal[0]))
tsp_graph = nx.Graph(tsp_edges)

nx.draw(tsp_graph, pos=vertices, with_labels=True)
```
